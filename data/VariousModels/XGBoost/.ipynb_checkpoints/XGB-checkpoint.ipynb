```python
import pickle
import numpy as np
import pandas as pd
import random
import xgboost as xgb
from xgboost import XGBRFClassifier
```

```python
# Location to the historical AFL match data
data_location = '../data'

# Filename of the pickled historical AFL match data
historical_data = pd.read_pickle("%s/historical_data_2007_2024.pkl"%(data_location))

# Convert these data types to numerical values
historical_data[['BrownlowVotes','CoachesVotes']] = historical_data[['BrownlowVotes','CoachesVotes']].apply(pd.to_numeric)
```

```python
# Grab the names of all the headers of the data
column_names = historical_data.columns.values.tolist()
# Assign indices to a list of equivalent length to the numbers of headers
column_indices = list(range(0, len(column_names)))

# Create a dictionary of header names for grabbing the correct statistical column if modifying which statistics to use
column_dict = {column_names[i]: column_indices[i] for i in range(len(column_names))}

# Names of columns (statistics) to use for the predictive model
stats_to_keep = ['NormDisposals', # Normalised disposals within a single match
                 'NormGoals', # Normalised number of goals within a single match
                 'NormTackles', # Normalised number of tackles within a single match
                 'NormMarks', # Normalised number of marks within a single match
                 'NormContestedMarks', # Normalised number of contested marks within a single match
                'NormInside50s', # Normalised number of inside 50s within a single match
                 'NormClearances', # Normalised number of clearances within a single match
                 'NormContestedPossessions', # Normalised number of contested possesions within a single match
                'NormGoalAssists', # Normalised number of goal assists within a single match
                 'NormDreamTeamPoints', # Normalised number of AFL dreamteam points within a single match
                 'NormSupercoach', # Normalised number of supercoach points within a single match
                 'Margin_y', # Final margin within a match. Found differences across two sources, but chose this one (name unhelpful!)
                 'CoachesVotes', # The awarded coaches votes for a single match
                 'Winner', # Binary variable denoting winner/loser of a single match
                 'Hitouts' # Number of hitouts in a single match
                ]

# Total number of features to be used for our classification model
n_features = len(stats_to_keep)
```

```python
# Determine the number of zero vote games to be used for the classification model
# Performance is sensitive to this choice, I found this to be a reasonable choice
zero_ratio = 5

# The total number of player statistics to use for the classification model
# The 3,2 and 1 Brownlow vote winners and the number of zero (3*zero_ratio) vote getters
number_of_players_for_classification = 3*(1 + zero_ratio)
```

```python
# Function to classify all the training data.
# Builds a reduced training set from all data.
# Designed to ensure all Brownlow vote getters and Coaches vote getters are included in the subset of training data

# Certainly could be structured better and cleaner (maybe will do that another time)

def classify_training_data(training_data_frame,individual_match_IDs,header_dict,kept_stats,                           
                          feature_num):
    # training_data_frame: the dataframe of the training data
    # individual_match_IDs: the individual match IDs signifying which matches to use
    # header_dict: dictionary of headers for selecting appropriate statistical information for the model
    # kept_stats: the names of the columns of statistics to be used in the classification model
    # feature_num: the number of features (statistics) to use for the model. This could be removed.

    # Returns:
    # full_stats_for_classification: array of player statistical information for training the classification model (model input)
    # brownlow_vote_for_classification: array of all Brownlow votes for the training data (used for predicting)

    
    # Determine the number of AFL games to be used in the training data
    n_games_considered = len(individual_match_IDs)

    # Allocate numpy array containing all the training data (statistics)
    full_stats_for_classification = np.zeros((n_games_training,number_of_players_for_classification,feature_num))
    # Allocate numpy array containing all the brownlow voting information
    brownlow_vote_for_classification = np.zeros((n_games_training,number_of_players_for_classification))

    # Loop over all the AFL games
    for i_games in range(n_games_considered):

        # Grab the information corresponding to the currect AFL match
        current_match = training_data_frame[(training_data_frame['MatchId'] == individual_match_IDs[i_games])].copy()    

        # Counter for storing player information in the training set
        player_counter = 0

        # Find the actual Brownlow vote winners for this match
        for i_vote in range(3):
            # Obtain the statistical information of the 3, 2 and 1 vote winner
            current_player = current_match[current_match['BrownlowVotes'] == i_vote+1].values.tolist()

            # Store the relevant (i.e. needed) statistical information to be used in our current classification model        
            for i_stat in range(feature_num):
                full_stats_for_classification[i_games][player_counter][i_stat] = current_player[0][header_dict[kept_stats[i_stat]]]

            # Store the Brownlow voting information
            brownlow_vote_for_classification[i_games][player_counter] = i_vote+1

            # increment current player number (this could certainly be done better)
            player_counter += 1

        if 'CoachesVotes' in kept_stats:
            # Grab all the players who received coaches votes for this particular match that did not receive Brownlow votes
            # Players with Brownlow votes were added above
            coach_vote_getters = current_match[(current_match['BrownlowVotes'] == 0) & (current_match['CoachesVotes'] > 0)].copy()

            # Loop through the selected player information
            for i_vote in range(len(coach_vote_getters)):
                current_player = coach_vote_getters.iloc[i_vote].values.tolist()
                # Add the full statistical information of the current player
                for i_stat in range(feature_num):
                    full_stats_for_classification[i_games][player_counter][i_stat] = current_player[header_dict[kept_stats[i_stat]]]
                
                # Continues to increment the number of current players used in the training set
                player_counter += 1

            n_coach_vote_getters = len(coach_vote_getters)
        else:
            n_coach_vote_getters = 0

        # Having added the player information of the Brownlow vote getters and Coaches Vote getters (if selected), fill the training
        # set with a random selection of players who did not receive any votes (Brownlow or Coach votes). Essentially, these are the
        # players for the zero classification
        if 'CoachesVotes' in kept_stats:
            zero_vote_getters = current_match[(current_match['BrownlowVotes'] == 0) & (current_match['CoachesVotes'] == 0)].copy()
        else:
            zero_vote_getters = current_match[(current_match['BrownlowVotes'] == 0)].copy()

        # Obtain indices of players for the zero classification
        num_indices = list(range(0, len(zero_vote_getters)))
        # Randomly select a subset of players
        random.shuffle(num_indices)

        # Fill the remaining classification training set with the appropriate number of zero vote getting players            
        for zero_vote in range(number_of_players_for_classification - 3 - n_coach_vote_getters):
            # Randomly selected player with zero votes
            current_player = zero_vote_getters.iloc[num_indices[zero_vote]].values.tolist()
            # Add corresponding vote information to the training set
            for i_stat in range(feature_num):
                full_stats_for_classification[i_games][player_counter][i_stat] = current_player[header_dict[kept_stats[i_stat]]]

            # Increment the player number for the training set
            player_counter += 1

    return full_stats_for_classification, brownlow_vote_for_classification
```

```python
# Function for providing AFL Brownlow vote predictions for all games within a single season

def season_prediction(data_for_forest, brownlow_information,
                      season_df, season_match_IDs,header_dict,kept_stats,feature_num):
    # Input:
    # data_for_forest: The statistical information for the classification model (input for predicting Brownlow votes)
    # brownlow_information: The historical Brownlow vote information for training the random forest classification model
    # season_df: the dataframe containing all the information for the current season to predict
    # season_match_IDs: the individual match IDs for the single season of matches
    # header_dict: dictionary of headers for selecting appropriate statistical information for the model
    # kept_stats: the names of the columns of statistics to be used in the classification model
    # feature_num: the number of features (statistics) to use for the model. This could be removed.

    # Returns:
    # brownlow_tally_predicted: list of all the predicted 3-2-1 votes by player for the provided AFL season
    # brownlow_tally_expected: list of the expected votes by player for the provided AFL season
    # totals_predicted: list containing a variety of summary information (number of games, number of correctly predicted votes etc.)
    
    # Number of games within the current season
    n_season_games = len(season_match_IDs)

    # An array for the stats of the current AFL player to predict
    normalised_stats = np.zeros(feature_num)

    # Number of players in a single match (can vary depending on various rules)
    match_players = int(len(season_df)/len(season_match_IDs))

    # Lists to contain the predicted (3-2-1 voting) and expected vote counts
    brownlow_tally_predicted = []
    brownlow_tally_expected = []

    # To contain various summary information
    totals_predicted = np.zeros(6)
    totals_predicted[0] = n_season_games

    # X and Y data for training the random forest
    X_train = data_for_forest
    y_train = brownlow_information

    # A random seed for seeding the initial values of the random forest
    rng_seed = int(10000*random.random())
    
    # Create the random forest model
    # Learning rate was the most important hyperparameter. A value of 2 was needed to obtain sensible expected votes
    rf = XGBRFClassifier(seed=rng_seed,n_estimators=200,colsample_bynode=0.9,colsample_bytree=0.9,max_depth=50,min_child_weight=5,
                        objective= 'binary:logistic',gamma=0,learning_rate=2)
    # Train the random forest model
    rf.fit(X_train, y_train)

    # Iterate over all AFL games within the season
    for i_games in range(n_season_games):

        # Lists to contain relevant expected vote information
        expected_vote_information = []

        # Obtain the current AFL match given the current match ID
        current_match = season_df[(season_df['MatchId'] == season_match_IDs[i_games])].copy()

        # The total expected votes for a match
        total_expected_votes = 0
        # Loop over all players within a single match
        for i_players in range(match_players):
            # Current AFL player under consideration
            current_player = current_match.iloc[i_players].values.tolist()

            # Obtain the relevant statistical information for the current player
            for i_stat in range(feature_num):
                normalised_stats[i_stat] = current_player[header_dict[kept_stats[i_stat]]]

            # Obtain the prediction probability for 0, 1, 2 and 3 Brownlow votes
            prediction_probability = rf.predict_proba(normalised_stats.reshape(1,-1))[0]
            
            expected_votes = 0
            # Calculate the total expected votes for the given player
            for i_vote in range(len(prediction_probability)):		
                # Sum of vote amount times the corresponding probability
                expected_votes += i_vote*prediction_probability[i_vote]

            # Store the expected vote information for the current match
            expected_vote_information.append([i_players,expected_votes])
            # Running total of expected votes within the match
            total_expected_votes += expected_votes

        # Obtain the normalisation for the expected votes within the match (must add up to 6)
        norm_total_expected_votes = 6./total_expected_votes
        
        # Sort the expected vote information for the match (to assign 3-2-1 votes to the 3 highest probability performances)
        sorted_expected_votes = sorted(expected_vote_information,key=lambda l:l[1],reverse=True)

        # Calculate the number of times 3, 2 and 1 votes are correctly assigned for the season
        # Note, all of these below could more easily be done with a confusion matrix
        if current_match['BrownlowVotes'].iloc[sorted_expected_votes[0][0]] == 3:
            totals_predicted[1] += 1
        if current_match['BrownlowVotes'].iloc[sorted_expected_votes[1][0]] == 2:
            totals_predicted[2] += 1
        if current_match['BrownlowVotes'].iloc[sorted_expected_votes[2][0]] == 1:
            totals_predicted[3] += 1

        # Number of times a player who actually receives 1 or 2 votes is predicted to get 1 or 2 votes
        # Used for exporing the accuracy when ordering not important
        if current_match['BrownlowVotes'].iloc[sorted_expected_votes[2][0]] == 1 or current_match['BrownlowVotes'].iloc[sorted_expected_votes[2][0]] == 2:
            totals_predicted[4] += 1        
        if current_match['BrownlowVotes'].iloc[sorted_expected_votes[1][0]] == 1 or current_match['BrownlowVotes'].iloc[sorted_expected_votes[1][0]] == 2:
            totals_predicted[4] += 1

        # Number of times a player is correctly identifed as receiving a vote (does not matter the amount)
        # Used for exporing accuracy of predicting the top three (how important is order)
        if current_match['BrownlowVotes'].iloc[sorted_expected_votes[0][0]] == 1 \
            or current_match['BrownlowVotes'].iloc[sorted_expected_votes[0][0]] == 2 \
            or current_match['BrownlowVotes'].iloc[sorted_expected_votes[0][0]] == 3:
                totals_predicted[5] += 1        
        if current_match['BrownlowVotes'].iloc[sorted_expected_votes[1][0]] == 1 \
            or current_match['BrownlowVotes'].iloc[sorted_expected_votes[1][0]] == 2 \
            or current_match['BrownlowVotes'].iloc[sorted_expected_votes[1][0]] == 3:
                totals_predicted[5] += 1        
        if current_match['BrownlowVotes'].iloc[sorted_expected_votes[2][0]] == 1 \
            or current_match['BrownlowVotes'].iloc[sorted_expected_votes[2][0]] == 2 \
            or current_match['BrownlowVotes'].iloc[sorted_expected_votes[2][0]] == 3:
                totals_predicted[5] += 1        

        # Assign 3-2-1 votes based on highest 3 expected votes for the given match
        for i in range(3):
            # Obtain current player name and team
            current_player = current_match['PlayerName'].iloc[sorted_expected_votes[i][0]]
            current_team = current_match['Team'].iloc[sorted_expected_votes[i][0]]

            # Flag for determining if the current player has already received votes in the season (for a different match)
            repeated_entry = False
            # Loop over all existing players to receive votes for the season
            for ii in range(len(brownlow_tally_predicted)):
                # Check if the current player already exists in our list of vote getters
                if current_player == brownlow_tally_predicted[ii][0] and current_team == brownlow_tally_predicted[ii][1]:
                    # If true, the current player already exists, so just add the votes the player got for this round
                    repeated_entry = True				
                    brownlow_tally_predicted[ii].append(3-i)

            # Current player does not yet have a vote for the season, create a new list entry for the current player
            if repeated_entry is False:
                brownlow_tally_predicted.append([current_player,current_team,3-i])

        # The same as above, but adding the expected votes for all players in the season
        for i_players in range(match_players):
            # Obtain current player name and team
            current_player = current_match['PlayerName'].iloc[sorted_expected_votes[i_players][0]]
            current_team = current_match['Team'].iloc[sorted_expected_votes[i_players][0]]

            # Flag for determining if the current player has already received votes in the season (for a different match)
            repeated_entry = False
            # Loop over all existing players to receive votes for the season
            for ii in range(len(brownlow_tally_expected)):
                # Check if the current player already exists in our list of vote getters
                if current_player == brownlow_tally_expected[ii][0] and current_team == brownlow_tally_expected[ii][1]:
                    # If true, the current player already exists, so just add the votes the player got for this round
                    repeated_entry = True				
                    brownlow_tally_expected[ii].append(float(sorted_expected_votes[i_players][1]*norm_total_expected_votes))

            # Current player does not yet have a vote for the season, create a new list entry for the current player
            if repeated_entry is False:
                brownlow_tally_expected.append([current_player,current_team,float(sorted_expected_votes[i_players][1]*norm_total_expected_votes)])

    return brownlow_tally_predicted, brownlow_tally_expected, totals_predicted
```

```python
# tuple of years corresponding to all available information
# A tuple for iterating over a loop as I switch the prediction year

training_years = ('2007','2008','2009','2010','2011','2012','2013','2014','2015','2016','2017'
                   ,'2018','2019','2020','2021','2022','2023','2024')

# For determining the current year from the above tuple
i_year = np.where(np.array(training_years) == '2024')[0][0]

# Create list of years for obtaining the correct training set data
new_training_years = list(training_years).copy()
new_training_years.remove(list(training_years)[i_year])
new_training_years = tuple(new_training_years)

print(new_training_years)
print(training_years[i_year])

# Obtain the relevant years of data for the training set
training_set = historical_data[historical_data['MatchId'].str.startswith(new_training_years)]
# Obtain the relevant year of data for predicting the AFL Brownlow
prediction_set = historical_data[historical_data['MatchId'].str.startswith(training_years[i_year])]

# Obtain the individual match IDs containined within the training set and the total number of matches in the training set
all_match_IDs_training = training_set['MatchId'].unique()
n_games_training = len(all_match_IDs_training)
# Obtain the individual match IDs for the season that we are predicting and the total number of matches in the season
all_match_IDs_prediction = prediction_set['MatchId'].unique()
n_games_prediction = len(all_match_IDs_prediction)

# Obtain the classified data for training the random forest model
full_stats_for_classification, brownlow_vote_for_classification = classify_training_data(
                                            training_set,all_match_IDs_training,column_dict,stats_to_keep,
                                            n_features)

# Reshape the information to be appropriate for input into the random classifier model
brownlow_information = np.reshape(brownlow_vote_for_classification,n_games_training*number_of_players_for_classification)
data_for_forest = np.reshape(full_stats_for_classification,(n_games_training*number_of_players_for_classification,n_features))

# Predict the current AFL season
brownlow_tally_predicted, brownlow_tally_expected, totals_predicted = season_prediction(
                                                                                                    data_for_forest, 
                                                                                                    brownlow_information,
                                                        prediction_set, all_match_IDs_prediction,column_dict,stats_to_keep,n_features)
    
```

```python
print('Predicted votes')
# Obtain a list of players and their total votes (awarded 3-2-1)
brownlow_count_predicted = []
for i in range(len(brownlow_tally_predicted)):
    # List of players, team and their total number of votes
	brownlow_count_predicted.append([brownlow_tally_predicted[i][0],brownlow_tally_predicted[i][1],sum(brownlow_tally_predicted[i][2:])])

# Create the tally of brownlow votes by sorting the total number of votes
sorted_tally_predicted = sorted(brownlow_count_predicted,key=lambda l:l[2],reverse=True)

# Output the top-10 votes
for ii in range(10):
	print(ii+1,sorted_tally_predicted[ii])
```

```python
print('Expected votes')
# Obtain a list of players and their total expected votes
brownlow_count_expected = []
for i in range(len(brownlow_tally_expected)):
    # List of players, team and their total number of votes
	brownlow_count_expected.append([brownlow_tally_expected[i][0],brownlow_tally_expected[i][1],sum(brownlow_tally_expected[i][2:])])

# Create the tally of brownlow votes by sorting the total number of votes
sorted_tally_expected = sorted(brownlow_count_expected,key=lambda l:l[2],reverse=True)

# Output the top-10 votes
for ii in range(10):
	print(ii+1,sorted_tally_expected[ii])
```
